```python
#checking the distribution of the Age column
df['Age'].value_counts()
```

```python
#filling the missing values in the Embarked column with the mode value of the Embarked column
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.isnull().sum()
```

```python
#filling missing values in the Age column with the median value of the Age column
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
#dropping the cabin column as it has too many missing values and is not useful for our analysis
df.drop('Cabin', axis=1, inplace=True)
```

```python

df.isnull().sum()
```

```python
#Survival count plot - How many people survived vs died.

plt.figure(figsize=(6,4))
sns.countplot(x='Survived', hue='Survived', data=df, palette='Set2', legend=False)
plt.title('Survival Count\n(0 = Died, 1 = Survived)')
plt.xlabel('Survived')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
#Survival by Sex : Difference in survival between males and females

plt.figure(figsize=(7,5))
sns.countplot(x='Sex', hue='Survived', data=df, palette='Set1')
plt.title('Survival by Sex')
plt.xlabel('Sex')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived', labels=['Died', 'Survived'])
plt.show()
```

```python
#Age distribution histogram : How ages are spread across the passengers

plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=30, kde=True, color='skyblue')
plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
#Age by survial: Whether younger or older passenger had better survival chances

plt.figure(figsize=(8,5))
sns.histplot(data=df, x='Age', hue='Survived', bins=30, kde=True, palette='coolwarm')
plt.title('Age Distribution by Survival Status')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
#Fare distribution by survival: Whether people who paid highter fares surviced more

plt.figure(figsize=(7,5))
sns.boxplot(x='Survived', y='Fare', hue='Survived', data=df, palette='pastel', legend=False)
plt.title('Fare Distribution by Survival')
plt.xlabel('Survived (0 = Died, 1 = Survived)')
plt.ylabel('Fare')
plt.show()
```

```python
#Heatmap, correlation Matrix: How strongly numerical features are related to each other 
plt.figure(figsize=(10,7))
numeric_df = df.select_dtypes(include=['int64', 'float64'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```
